```python
import pandas as pd
import json
import random
 
def excel_to_json_finetuning(excel_path, output_json_path, samples_per_column=13, repeats_per_column=3):
    # Read the Excel file
    df = pd.read_csv(excel_path)
    # Get unique column names (case-sensitive)
    unique_columns = list(df.columns)
    # Prepare the output data
    output_data = []
    for column in unique_columns:
        # Get non-null values from the column
        non_null_values = df[column].dropna().astype(str).tolist()
        # Skip if no values in the column or not enough for samples
        if len(non_null_values) < samples_per_column:
            continue
        # Create multiple entries for each unique column (N times)
        for _ in range(repeats_per_column):
            # Get new random samples each time
            sample_values = random.sample(non_null_values, samples_per_column)
            # Create the JSON structure
            entry = {
                "instruction": f"Explain the function of the {column} column in customer-related data.",
                "input": {
                    "column_name": column,
                    "sample_values": sample_values
                },
                "answer": f"The {column} column stores"
            }
            output_data.append(entry)
    # Save to JSON file
    with open(output_json_path, 'w', encoding='utf-8') as f:
        json.dump(output_data, f, indent=2)
    return output_data
 
# Example usage
excel_file_path = 'output_data.csv'  # Replace with your Excel file path
output_json_path = 'finetuning_data.json'
result = excel_to_json_finetuning(excel_file_path, output_json_path, repeats_per_column=3)
```
